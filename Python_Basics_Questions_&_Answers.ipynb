## 26. Body Mass Index (BMI) Calculation

```python
weight = float(input("Enter weight in kg: "))
height = float(input("Enter height in meters: "))

bmi = weight / (height ** 2)
print("BMI:", round(bmi, 2))

if bmi < 18.5:
    print("Underweight")
elif bmi < 25:
    print("Normal weight")
elif bmi < 30:
    print("Overweight")
else:
    print("Obese")
```

## 33. Check for Prime Number

```python
n = int(input("Enter a number: "))

if n < 2:
    print("Not a prime number")
else:
    prime = True
    for i in range(2, n):
        if n % i == 0:
            prime = False
            break
    if prime:
        print("Prime number")
    else:
        print("Not a prime number")
```

## 34. Count Vowels in a String

## 42. Simple Number Guessing Game

```python
secret_number = 7

guess = int(input("Guess the secret number: "))

while guess != secret_number:
    if guess < secret_number:
        print("Too low")
    else:
        print("Too high")
    guess = int(input("Try again: "))

print("Correct! You guessed the number.")
```

## 43. Reverse a Number

```python
number = int(input("Enter a number: "))
reverse = 0

while number > 0:
    digit = number % 10
    reverse = reverse * 10 + digit
    number //= 10

print("Reversed number:", reverse)
```
