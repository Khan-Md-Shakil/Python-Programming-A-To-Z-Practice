## 1. Functions
A function is a block of code that performs a specific task.

### Create a Function
**Question/Example:** Create a function named `greet()` that prints `Hello World!`.

````python
def greet():
    print("Hello World!")
````

### Calling a Function
**Question/Example:** Define `greet()`, call the function, and print a message outside the function.

````python
def greet():
    print("Hello World!")

greet()

print("Outside Function")
````

### Function Arguments
**Question/Example:** Create a function to add two numbers using two arguments.

````python
def add_numbers(num1, num2):
    sum = num1 + num2
    print("Sum:", sum)

add_numbers(5, 4)
````

### Return Statement
**Question/Example:** Return a value from a function using `return`.

````python
def add_numbers(num1, num2):
    return num1 + num2

result = add_numbers(5, 4)
print("Sum:", result)
````

### Default Arguments in Functions
**Question/Example:** Create `greet()` with a default message.

````python
def greet(name, message="Hello"):
    print(message, name)

greet("Alice", "Good Morning")
greet("Bob")
````

### *args in Functions
**Question/Example:** Use `*args` to add any number of positional arguments.

````python
def add_all(*numbers):
    return sum(numbers)

print(add_all(1, 2, 3, 4))
````

### **kwargs in Functions
**Question/Example:** Use `**kwargs` to accept any number of keyword arguments.

````python
def greet(**words):
    for key, value in words.items():
        print(f"{key}: {value}")

greet(name="John", greeting="Hello")
````

### Local Variable
**Question/Example:** Demonstrate a variable declared inside a function.

````python
def show_number():
    number = 10
    print("Local variable:", number)

show_number()
````

### Global Variable
**Question/Example:** Demonstrate a variable declared outside a function.

````python
number = 10

def show_number():
    print("Global variable:", number)

show_number()
print("Outside function:", number)
````

### Nonlocal Variable
**Question/Example:** Demonstrate the `nonlocal` keyword in a nested function.

````python
def outer():
    number = 10

    def inner():
        nonlocal number
        number += 5
        print("Inside inner:", number)

    inner()
    print("Inside outer:", number)

outer()
````

### Changing a Global Variable Using `global`
**Question/Example:** Modify a global variable from inside a function.

````python
number = 10

def change_number():
    global number
    number = 20

change_number()
print(number)
````

### Function — Check if a Number is Prime
**Question:** Write a function to check whether a number is prime.

````python
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True

number = 29
print(f"Is {number} prime? {is_prime(number)}")
````

### Function — Greatest Common Divisor (GCD)
**Question:** Write a function to calculate the GCD of two numbers.

````python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a

num1, num2 = 48, 18
print(f"GCD of {num1} and {num2} is {gcd(num1, num2)}")
````

### Function — Sum of Digits of a Number
**Question:** Write a function to find the sum of the digits of a number.

````python
def sum_of_digits(n):
    total = 0

    while n > 0:
        total += n % 10
        n //= 10

    return total

number = 1234
print(f"Sum of digits of {number} is {sum_of_digits(number)}")
````

### Function — Longest Word in a Sentence
**Question:** Write a function to find the longest word in a sentence.

````python
def longest_word(sentence):
    words = sentence.split()
    longest = max(words, key=len)
    return longest

sentence = "The quick brown fox jumps over the lazy dog"
print(f"The longest word in the sentence is: {longest_word(sentence)}")
````

### Function — Area of a Circle
**Question:** Write a function to calculate the area of a circle.

````python
import math

def area_of_circle(radius):
    return math.pi * (radius ** 2)

radius = 5
print(f"Area of the circle with radius {radius} is: {area_of_circle(radius)}")
````

## Function Practice Questions

**H1:** Write a function to calculate the sum of numbers from 1 to N.

**H2:** Write a function that returns the square of a number.

**H3:** Write a function that checks whether a number is even/odd.

**H4:** Write a function that returns the larger number from two given numbers.

````python
# H1
def sum_to_n(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

print(sum_to_n(10))
````

````python
# H2
def square(number):
    return number ** 2

print(square(5))
````

````python
# H3
def even_or_odd(number):
    if number % 2 == 0:
        return "Even"
    return "Odd"

print(even_or_odd(7))
````

````python
# H4
def larger(a, b):
    if a > b:
        return a
    return b

print(larger(15, 20))
````

### Merge Two Lists and Remove Duplicates

````python
list1 = input("Enter elements of the first list separated by spaces: ").split()
list1 = [int(x) for x in list1]

list2 = input("Enter elements of the second list separated by spaces: ").split()
list2 = [int(x) for x in list2]

merged_list = list1 + list2
merged_list = list(set(merged_list))

print("Merged list without duplicates:", merged_list)
````

### Find the Intersection of Two Lists

````python
list1 = input("Enter elements of the first list separated by spaces: ").split()
list1 = [int(x) for x in list1]

list2 = input("Enter elements of the second list separated by spaces: ").split()
list2 = [int(x) for x in list2]

intersection = list(set(list1) & set(list2))

print("Intersection of the two lists:", intersection)
````

### Create a Dictionary

### Dictionary Keys Must Be Immutable and Unique

````python
example = {
    "name": "John",
    1: "One",
    (2, 3): "Tuple key"
}

print(example)

numbers = {"a": 10, "a": 20}
print(numbers)
````

### Access Dictionary Items Using a Key

````python
country_capitals = {
    "Germany": "Berlin",
    "Italy": "Rome",
    "England": "London"
}

print(country_capitals["Italy"])
````

### Access Dictionary Items Using `get()`

````python
country_capitals = {
    "Germany": "Berlin",
    "Italy": "Rome",
    "England": "London"
}

print(country_capitals.get("Italy"))
print(country_capitals.get("France", "Key not found"))
````

### Remove Dictionary Item with `del`

````python
country_capitals = {
    "Germany": "Berlin",
    "Italy": "Rome",
    "England": "London"
}

del country_capitals["Italy"]
print(country_capitals)
````

### Remove Dictionary Item with `pop()`

````python
country_capitals = {
    "Germany": "Berlin",
    "Italy": "Rome",
    "England": "London"
}

removed = country_capitals.pop("Italy")
print("Removed:", removed)
print(country_capitals)
````

### Remove All Dictionary Items with `clear()`

````python
country_capitals = {
    "Germany": "Berlin",
    "Italy": "Rome",
    "England": "London"
}

country_capitals.clear()
print(country_capitals)
````

### Change Dictionary Items

````python
country_capitals = {
    "Germany": "Berlin",
    "Italy": "Naples",
    "England": "London"
}

country_capitals["Italy"] = "Rome"
print(country_capitals)
````

### Add/Change Dictionary Items with `update()`

````python
marks = {"Physics": 67, "Maths": 87}
internal_marks = {"Practical": 48}

marks.update(internal_marks)

print(marks)
````

### Iterate Through a Dictionary

````python
country_capitals = {
    "United States": "Washington D.C.",
    "Italy": "Rome"
}

for country in country_capitals:
    print(country)

print()

for country in country_capitals:
    capital = country_capitals[country]
    print(capital)
````

### Find Dictionary Length

````python
country_capitals = {"England": "London", "Italy": "Rome"}
print(len(country_capitals))

numbers = {10: "ten", 20: "twenty", 30: "thirty"}
print(len(numbers))

countries = {}
print(len(countries))
````

### Dictionary Methods

````python
student = {
    "name": "John",
    "age": 20,
    "department": "CSE"
}

print(student.keys())
print(student.values())
print(student.get("name"))
print(student.copy())

removed = student.pop("age")
print("Removed:", removed)
print(student)

student.update({"city": "Nantong"})
print(student)

last_item = student.popitem()
print("Last item:", last_item)
print(student)

student.clear()
print(student)
````
